```python
# ================================================================
# STUDENT PLACEMENT PREDICTION SYSTEM
# PHASE 1 — COMPLETE ML PIPELINE
# ================================================================
#
# Pipeline:
# Dataset
#   ↓
# Data Inspection
#   ↓
# Data Cleaning
#   ↓
# EDA
#   ↓
# Target Preparation
#   ↓
# Leakage Removal
#   ↓
# Train/Test Split
#   ↓
# Preprocessing Pipeline
#   ↓
# Model Training
#   ├── Logistic Regression
#   ├── Random Forest
#   └── XGBoost
#   ↓
# Model Comparison
#   ↓
# Hyperparameter Tuning
#   ↓
# Final Model
#   ↓
# Evaluation
#   ↓
# SHAP Explainability
#   ↓
# Save Complete Pipeline
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import os
import warnings
import joblib
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RandomizedSearchCV
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    LabelEncoder
)

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

from sklearn.utils.multiclass import type_of_target

warnings.filterwarnings("ignore")

print("All basic libraries imported successfully!")


# ================================================================
# 2. IMPORT XGBOOST
# ================================================================

try:
    from xgboost import XGBClassifier
    XGBOOST_AVAILABLE = True
    print("XGBoost imported successfully!")
except ImportError:
    XGBOOST_AVAILABLE = False
    print("XGBoost is not installed.")
    print("Install it using:")
    print("pip install xgboost")


# ================================================================
# 3. CONFIGURATION
# ================================================================
#
# CHANGE THESE TWO VALUES
#
# Example:
# DATA_PATH = "student_placement.csv"
# TARGET_COLUMN = "placement_status"
#
# If your dataset uses another target column name,
# replace it accordingly.
# ================================================================

DATA_PATH = r"C:\Users\ASUS\My_Learning\MLops project\students.csv"

TARGET_COLUMN = "Placement Status"

# Folder for saving all outputs
MODEL_DIR = "models"
REPORT_DIR = "reports"
FIGURE_DIR = os.path.join(REPORT_DIR, "figures")

os.makedirs(MODEL_DIR, exist_ok=True)
os.makedirs(REPORT_DIR, exist_ok=True)
os.makedirs(FIGURE_DIR, exist_ok=True)

print("Configuration completed.")
print("Dataset :", DATA_PATH)
print("Target  :", TARGET_COLUMN)


# ================================================================
# 4. LOAD DATASET
# ================================================================

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"\nDataset not found: {DATA_PATH}\n"
        "Place the CSV file in the same folder as the notebook "
        "or provide the complete path."
    )

df = pd.read_csv(DATA_PATH)

print("\nDataset loaded successfully!")
print("=" * 60)
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
print("=" * 60)


# ================================================================
# 5. DISPLAY FIRST RECORDS
# ================================================================

print("\nFirst 5 records:")
display(df.head())


# ================================================================
# 6. DISPLAY COLUMN NAMES
# ================================================================

print("\nDataset Columns:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")


# ================================================================
# 7. DATASET INFORMATION
# ================================================================

print("\nDataset Information:")
df.info()


# ================================================================
# 8. STATISTICAL SUMMARY
# ================================================================

print("\nStatistical Summary:")
display(df.describe(include="all").T)


# ================================================================
# 9. CHECK DUPLICATES
# ================================================================

duplicate_count = df.duplicated().sum()

print("\nDuplicate rows:", duplicate_count)

if duplicate_count > 0:
    print("Removing duplicate rows...")
    df = df.drop_duplicates().reset_index(drop=True)
    print("Rows after duplicate removal:", len(df))


# ================================================================
# 10. CHECK MISSING VALUES
# ================================================================

missing_table = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().sum().values / len(df) * 100
    )
})

missing_table = missing_table.sort_values(
    by="Missing Values",
    ascending=False
)

print("\nMissing Value Report:")
display(missing_table)


# ================================================================
# 11. CHECK DATA TYPES
# ================================================================

dtype_table = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Unique Values": [
        df[col].nunique(dropna=True)
        for col in df.columns
    ]
})

print("\nData Type Report:")
display(dtype_table)


# ================================================================
# 12. VERIFY TARGET COLUMN
# ================================================================

if TARGET_COLUMN not in df.columns:
    print("\nERROR: Target column was not found.")
    print("\nAvailable columns:")
    print(df.columns.tolist())

    raise ValueError(
        f"\nChange TARGET_COLUMN to the correct column name."
    )

print("\nTarget column found:", TARGET_COLUMN)


# ================================================================
# 13. TARGET VALUE ANALYSIS
# ================================================================

print("\nTarget values:")
print(df[TARGET_COLUMN].value_counts(dropna=False))

print("\nTarget percentages:")
print(
    df[TARGET_COLUMN]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)


# ================================================================
# 14. REMOVE ROWS WITH MISSING TARGET
# ================================================================

target_missing = df[TARGET_COLUMN].isnull().sum()

if target_missing > 0:
    print(
        f"\nRemoving {target_missing} rows "
        "with missing target values."
    )

    df = df.dropna(
        subset=[TARGET_COLUMN]
    ).reset_index(drop=True)


# ================================================================
# 15. TARGET DISTRIBUTION PLOT
# ================================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x=TARGET_COLUMN
)

plt.title("Placement Target Distribution")
plt.xlabel(TARGET_COLUMN)
plt.ylabel("Number of Students")
plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "target_distribution.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 16. IDENTIFY NUMERICAL AND CATEGORICAL FEATURES
# ================================================================

X_temp = df.drop(columns=[TARGET_COLUMN])

numerical_columns = X_temp.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = X_temp.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumerical Features:")
for col in numerical_columns:
    print(" -", col)

print("\nCategorical Features:")
for col in categorical_columns:
    print(" -", col)


# ================================================================
# 17. NUMERICAL FEATURE DISTRIBUTIONS
# ================================================================

if len(numerical_columns) > 0:

    df[numerical_columns].hist(
        figsize=(16, 12),
        bins=20
    )

    plt.suptitle(
        "Numerical Feature Distributions",
        fontsize=16
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            FIGURE_DIR,
            "numerical_distributions.png"
        ),
        dpi=300
    )

    plt.show()


# ================================================================
# 18. CORRELATION MATRIX
# ================================================================

if len(numerical_columns) > 1:

    plt.figure(figsize=(14, 10))

    correlation_matrix = df[
        numerical_columns
    ].corr()

    sns.heatmap(
        correlation_matrix,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0
    )

    plt.title("Numerical Feature Correlation Matrix")

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            FIGURE_DIR,
            "correlation_matrix.png"
        ),
        dpi=300
    )

    plt.show()


# ================================================================
# 19. DATA LEAKAGE / UNWANTED COLUMN REMOVAL
# ================================================================
#
# IMPORTANT:
# These are examples of columns that should generally NOT be
# used to predict whether a student gets placed if the information
# is only available after placement.
#
# The code only removes a column if that exact column exists.
# ================================================================

POSSIBLE_ID_COLUMNS = [
    "student_id",
    "studentid",
    "student id",
    "id",
    "serial_no",
    "serial number",
    "s_no",
    "sno"
]

POSSIBLE_LEAKAGE_COLUMNS = [
    "salary",
    "ctc",
    "package",
    "salary_lpa",
    "company",
    "company_name",
    "company_name",
    "placement_date",
    "joining_date",
    "offer_date",
    "offer_letter",
    "job_offer",
    "offer_status",
    "joining_status"
]

columns_to_remove = []

for col in df.columns:

    col_lower = col.lower().strip()

    if col_lower in POSSIBLE_ID_COLUMNS:
        if col != TARGET_COLUMN:
            columns_to_remove.append(col)

    if col_lower in POSSIBLE_LEAKAGE_COLUMNS:
        if col != TARGET_COLUMN:
            columns_to_remove.append(col)


columns_to_remove = list(set(columns_to_remove))

print("\nPotential ID / leakage columns detected:")

if columns_to_remove:
    for col in columns_to_remove:
        print(" -", col)
else:
    print("None automatically detected.")


# ================================================================
# 20. REMOVE IDENTIFIED COLUMNS
# ================================================================

if columns_to_remove:

    df = df.drop(
        columns=columns_to_remove
    )

    print("\nRemoved columns:")
    print(columns_to_remove)


# ================================================================
# 21. RECHECK FEATURES
# ================================================================

X = df.drop(
    columns=[TARGET_COLUMN]
)

y = df[TARGET_COLUMN]

print("\nFeature matrix shape :", X.shape)
print("Target shape         :", y.shape)


# ================================================================
# 22. CONVERT TARGET TO BINARY
# ================================================================
#
# This section attempts to convert common placement labels into:
#
# Not Placed = 0
# Placed     = 1
#
# If your target is already 0/1, it remains unchanged.
# ================================================================

print("\nOriginal target values:")
print(y.value_counts())


def convert_target(value):
    """
    Convert common placement labels into binary values.
    """

    if pd.isna(value):
        return np.nan

    value_str = str(value).strip().lower()

    # Positive labels
    positive_labels = [
        "placed",
        "yes",
        "y",
        "true",
        "1",
        "selected",
        "placement",
        "successful"
    ]

    # Negative labels
    negative_labels = [
        "not placed",
        "not_placed",
        "notplaced",
        "no",
        "n",
        "false",
        "0",
        "not selected",
        "unsuccessful"
    ]

    if value_str in positive_labels:
        return 1

    if value_str in negative_labels:
        return 0

    return value


y_converted = y.apply(convert_target)

unique_target_values = y_converted.dropna().unique()

print("\nTarget values after conversion:")
print(unique_target_values)


# ================================================================
# 23. ENCODE TARGET IF STILL CATEGORICAL
# ================================================================

if not set(unique_target_values).issubset({0, 1}):

    print(
        "\nTarget contains labels that were not automatically "
        "recognized as binary."
    )

    print("Encoding target using LabelEncoder.")

    label_encoder = LabelEncoder()

    y = pd.Series(
        label_encoder.fit_transform(
            y_converted.astype(str)
        ),
        index=y_converted.index,
        name=TARGET_COLUMN
    )

    print("\nTarget mapping:")

    for original, encoded in zip(
        label_encoder.classes_,
        label_encoder.transform(
            label_encoder.classes_
        )
    ):
        print(
            f"{original} -> {encoded}"
        )

else:

    y = y_converted.astype(int)


# ================================================================
# 24. FINAL TARGET CHECK
# ================================================================

print("\nFinal target distribution:")
print(y.value_counts())

print("\nFinal target percentages:")
print(
    y.value_counts(normalize=True)
    .mul(100)
    .round(2)
)


# ================================================================
# 25. REMOVE CONSTANT FEATURES
# ================================================================
#
# A feature with only one unique value provides no useful
# information to a classification model.
# ================================================================

constant_columns = [
    col for col in X.columns
    if X[col].nunique(dropna=False) <= 1
]

if constant_columns:

    print("\nRemoving constant columns:")

    for col in constant_columns:
        print(" -", col)

    X = X.drop(
        columns=constant_columns
    )


# ================================================================
# 26. REIDENTIFY FEATURE TYPES
# ================================================================

numerical_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nFinal Numerical Features:")
print(numerical_features)

print("\nFinal Categorical Features:")
print(categorical_features)


# ================================================================
# 27. TRAIN / TEST SPLIT
# ================================================================
#
# 80% development
# 20% final untouched test
#
# Stratification maintains the target class distribution.
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nDataset Split:")
print("----------------------------")
print("Training samples :", len(X_train))
print("Testing samples  :", len(X_test))
print("----------------------------")


# ================================================================
# 28. BUILD PREPROCESSING PIPELINE
# ================================================================
#
# Numerical:
# Missing values → Median → StandardScaler
#
# Categorical:
# Missing values → Most Frequent → OneHotEncoder
# ================================================================

numerical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)


transformers = []

if numerical_features:

    transformers.append(
        (
            "numerical",
            numerical_transformer,
            numerical_features
        )
    )


if categorical_features:

    transformers.append(
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    )


preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)


print("\nPreprocessing pipeline created.")


# ================================================================
# 29. DEFINE MODELS
# ================================================================

models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=2000,
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            n_jobs=-1,
            class_weight="balanced"
        )
}


# Add XGBoost if installed

if XGBOOST_AVAILABLE:

    models["XGBoost"] = XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )


# ================================================================
# 30. TRAIN BASELINE MODELS
# ================================================================

results = []

trained_pipelines = {}


for model_name, model in models.items():

    print("\n" + "=" * 70)
    print(f"TRAINING: {model_name}")
    print("=" * 70)

    pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "model",
                model
            )
        ]
    )

    pipeline.fit(
        X_train,
        y_train
    )

    trained_pipelines[
        model_name
    ] = pipeline

    # Predictions
    y_pred = pipeline.predict(
        X_test
    )

    # Probability
    try:
        y_probability = pipeline.predict_proba(
            X_test
        )[:, 1]

        roc_auc = roc_auc_score(
            y_test,
            y_probability
        )

    except Exception:
        y_probability = None
        roc_auc = np.nan

    # Metrics
    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": roc_auc
    })

    print(
        f"Accuracy  : {accuracy:.4f}"
    )

    print(
        f"Precision : {precision:.4f}"
    )

    print(
        f"Recall    : {recall:.4f}"
    )

    print(
        f"F1 Score  : {f1:.4f}"
    )

    print(
        f"ROC-AUC   : {roc_auc:.4f}"
    )


# ================================================================
# 31. MODEL COMPARISON
# ================================================================

results_df = pd.DataFrame(
    results
)

results_df = results_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)


# Save model comparison

results_df.to_csv(
    os.path.join(
        REPORT_DIR,
        "model_comparison.csv"
    ),
    index=False
)


# ================================================================
# 32. CONFUSION MATRIX FOR EACH MODEL
# ================================================================

for model_name, pipeline in trained_pipelines.items():

    y_pred = pipeline.predict(
        X_test
    )

    cm = confusion_matrix(
        y_test,
        y_pred
    )

    plt.figure(figsize=(6, 5))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=[
            "Not Placed",
            "Placed"
        ],
        yticklabels=[
            "Not Placed",
            "Placed"
        ]
    )

    plt.title(
        f"Confusion Matrix - {model_name}"
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.tight_layout()

    safe_name = (
        model_name
        .lower()
        .replace(" ", "_")
    )

    plt.savefig(
        os.path.join(
            FIGURE_DIR,
            f"confusion_matrix_{safe_name}.png"
        ),
        dpi=300
    )

    plt.show()


# ================================================================
# 33. ROC CURVES
# ================================================================

plt.figure(figsize=(9, 7))

for model_name, pipeline in trained_pipelines.items():

    try:

        probabilities = pipeline.predict_proba(
            X_test
        )[:, 1]

        fpr, tpr, _ = roc_curve(
            y_test,
            probabilities
        )

        auc_score = roc_auc_score(
            y_test,
            probabilities
        )

        plt.plot(
            fpr,
            tpr,
            label=f"{model_name} (AUC={auc_score:.3f})"
        )

    except Exception:
        pass


plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "ROC Curves - Model Comparison"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "roc_curves.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 34. SELECT MODEL FOR TUNING
# ================================================================
#
# We use F1 Score as the initial comparison criterion.
# For an imbalanced dataset, this is generally more informative
# than accuracy alone.
# ================================================================

best_baseline_name = results_df.iloc[0]["Model"]

print(
    "\nBest baseline model based on F1 Score:",
    best_baseline_name
)


# ================================================================
# 35. HYPERPARAMETER TUNING
# ================================================================

best_pipeline = None
best_model_name = None

if best_baseline_name == "Random Forest":

    print("\nStarting Random Forest hyperparameter tuning...")

    rf_pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "model",
                RandomForestClassifier(
                    random_state=42,
                    n_jobs=-1,
                    class_weight="balanced"
                )
            )
        ]
    )

    param_grid = {

        "model__n_estimators": [
            200,
            300,
            500
        ],

        "model__max_depth": [
            None,
            5,
            10,
            20
        ],

        "model__min_samples_split": [
            2,
            5,
            10
        ],

        "model__min_samples_leaf": [
            1,
            2,
            4
        ],

        "model__max_features": [
            "sqrt",
            "log2"
        ]
    }

    search = RandomizedSearchCV(
        estimator=rf_pipeline,
        param_distributions=param_grid,
        n_iter=20,
        scoring="f1",
        cv=StratifiedKFold(
            n_splits=5,
            shuffle=True,
            random_state=42
        ),
        random_state=42,
        n_jobs=-1,
        verbose=1
    )

    search.fit(
        X_train,
        y_train
    )

    best_pipeline = search.best_estimator_
    best_model_name = "Tuned Random Forest"

    print("\nBest parameters:")
    print(search.best_params_)

    print(
        "\nBest cross-validation F1:",
        search.best_score_
    )


elif best_baseline_name == "XGBoost" and XGBOOST_AVAILABLE:

    print("\nStarting XGBoost hyperparameter tuning...")

    xgb_pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "model",
                XGBClassifier(
                    objective="binary:logistic",
                    eval_metric="logloss",
                    random_state=42,
                    n_jobs=-1
                )
            )
        ]
    )

    param_grid = {

        "model__n_estimators": [
            200,
            300,
            500
        ],

        "model__max_depth": [
            3,
            4,
            5,
            6,
            8
        ],

        "model__learning_rate": [
            0.01,
            0.03,
            0.05,
            0.1
        ],

        "model__subsample": [
            0.7,
            0.8,
            1.0
        ],

        "model__colsample_bytree": [
            0.7,
            0.8,
            1.0
        ]
    }

    search = RandomizedSearchCV(
        estimator=xgb_pipeline,
        param_distributions=param_grid,
        n_iter=20,
        scoring="f1",
        cv=StratifiedKFold(
            n_splits=5,
            shuffle=True,
            random_state=42
        ),
        random_state=42,
        n_jobs=-1,
        verbose=1
    )

    search.fit(
        X_train,
        y_train
    )

    best_pipeline = search.best_estimator_
    best_model_name = "Tuned XGBoost"

    print("\nBest parameters:")
    print(search.best_params_)

    print(
        "\nBest cross-validation F1:",
        search.best_score_
    )


else:

    print(
        "\nBest baseline is Logistic Regression."
    )

    print(
        "Performing Logistic Regression tuning..."
    )

    lr_pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "model",
                LogisticRegression(
                    max_iter=3000,
                    random_state=42
                )
            )
        ]
    )

    param_grid = {

        "model__C": [
            0.01,
            0.1,
            1,
            10,
            100
        ],

        "model__solver": [
            "liblinear",
            "lbfgs"
        ]
    }

    search = RandomizedSearchCV(
        estimator=lr_pipeline,
        param_distributions=param_grid,
        n_iter=10,
        scoring="f1",
        cv=StratifiedKFold(
            n_splits=5,
            shuffle=True,
            random_state=42
        ),
        random_state=42,
        n_jobs=-1,
        verbose=1
    )

    search.fit(
        X_train,
        y_train
    )

    best_pipeline = search.best_estimator_
    best_model_name = "Tuned Logistic Regression"

    print("\nBest parameters:")
    print(search.best_params_)

    print(
        "\nBest cross-validation F1:",
        search.best_score_
    )


# ================================================================
# 36. FINAL MODEL EVALUATION
# ================================================================

print("\n" + "=" * 70)
print("FINAL MODEL EVALUATION")
print("=" * 70)

y_pred_final = best_pipeline.predict(
    X_test
)

y_probability_final = best_pipeline.predict_proba(
    X_test
)[:, 1]


final_accuracy = accuracy_score(
    y_test,
    y_pred_final
)

final_precision = precision_score(
    y_test,
    y_pred_final,
    zero_division=0
)

final_recall = recall_score(
    y_test,
    y_pred_final,
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    y_pred_final,
    zero_division=0
)

final_roc_auc = roc_auc_score(
    y_test,
    y_probability_final
)


print("\nFinal Model:", best_model_name)

print(
    f"\nAccuracy  : {final_accuracy:.4f}"
)

print(
    f"Precision : {final_precision:.4f}"
)

print(
    f"Recall    : {final_recall:.4f}"
)

print(
    f"F1 Score  : {final_f1:.4f}"
)

print(
    f"ROC-AUC   : {final_roc_auc:.4f}"
)


# ================================================================
# 37. CLASSIFICATION REPORT
# ================================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_final,
        target_names=[
            "Not Placed",
            "Placed"
        ],
        zero_division=0
    )
)


# ================================================================
# 38. FINAL CONFUSION MATRIX
# ================================================================

final_cm = confusion_matrix(
    y_test,
    y_pred_final
)

plt.figure(figsize=(7, 6))

sns.heatmap(
    final_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Not Placed",
        "Placed"
    ],
    yticklabels=[
        "Not Placed",
        "Placed"
    ]
)

plt.title(
    f"Final Model - {best_model_name}"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "final_confusion_matrix.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 39. FINAL ROC CURVE
# ================================================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability_final
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {final_roc_auc:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "Final Model ROC Curve"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "final_roc_curve.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 40. SHAP EXPLAINABILITY
# ================================================================

print("\nAttempting SHAP explainability...")

try:

    import shap

    print("SHAP imported successfully.")

    # Get trained model
    fitted_model = best_pipeline.named_steps["model"]

    # Transform test data
    fitted_preprocessor = (
        best_pipeline.named_steps["preprocessor"]
    )

    X_test_transformed = (
        fitted_preprocessor.transform(X_test)
    )

    # Get feature names
    try:

        feature_names = (
            fitted_preprocessor
            .get_feature_names_out()
        )

    except Exception:

        feature_names = [
            f"feature_{i}"
            for i in range(
                X_test_transformed.shape[1]
            )
        ]

    # Convert sparse matrix if required
    if hasattr(
        X_test_transformed,
        "toarray"
    ):

        X_test_shap = (
            X_test_transformed.toarray()
        )

    else:

        X_test_shap = X_test_transformed


    # Tree-based models
    if (
        "RandomForest" in best_model_name
        or "XGBoost" in best_model_name
    ):

        explainer = shap.TreeExplainer(
            fitted_model
        )

        shap_values = explainer.shap_values(
            X_test_shap
        )

        # Handle different SHAP versions
        if isinstance(
            shap_values,
            list
        ):

            shap_plot_values = shap_values[-1]

        elif (
            hasattr(
                shap_values,
                "ndim"
            )
            and shap_values.ndim == 3
        ):

            shap_plot_values = (
                shap_values[:, :, -1]
            )

        else:

            shap_plot_values = shap_values


        plt.figure()

        shap.summary_plot(
            shap_plot_values,
            X_test_shap,
            feature_names=feature_names,
            show=False
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                FIGURE_DIR,
                "shap_summary.png"
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.show()

    else:

        # Linear model
        explainer = shap.LinearExplainer(
            fitted_model,
            X_test_shap
        )

        shap_values = explainer(
            X_test_shap
        )

        plt.figure()

        shap.summary_plot(
            shap_values,
            X_test_shap,
            feature_names=feature_names,
            show=False
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                FIGURE_DIR,
                "shap_summary.png"
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.show()


except ImportError:

    print(
        "\nSHAP is not installed."
    )

    print(
        "Install it using:"
    )

    print(
        "pip install shap"
    )

except Exception as e:

    print(
        "\nSHAP could not be generated."
    )

    print(
        "Reason:",
        str(e)
    )


# ================================================================
# 41. SAVE FINAL MODEL PIPELINE
# ================================================================
#
# IMPORTANT:
# We save the COMPLETE pipeline:
#
# Raw student data
#      ↓
# Preprocessing
#      ↓
# Encoding
#      ↓
# Scaling
#      ↓
# Model
#
# This is much better than saving only the classifier.
# ================================================================

MODEL_PATH = os.path.join(
    MODEL_DIR,
    "placement_pipeline.pkl"
)

joblib.dump(
    best_pipeline,
    MODEL_PATH
)

print("\n" + "=" * 70)
print("MODEL SAVED")
print("=" * 70)

print(
    f"\nSaved model path:\n{MODEL_PATH}"
)


# ================================================================
# 42. SAVE FINAL METRICS
# ================================================================

final_metrics = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],

    "Value": [
        final_accuracy,
        final_precision,
        final_recall,
        final_f1,
        final_roc_auc
    ]
})

display(
    final_metrics.style.format({
        "Value": "{:.4f}"
    })
)

final_metrics.to_csv(
    os.path.join(
        REPORT_DIR,
        "final_metrics.csv"
    ),
    index=False
)


# ================================================================
# 43. SAVE FEATURE INFORMATION
# ================================================================

feature_information = {
    "target_column": TARGET_COLUMN,
    "numerical_features": numerical_features,
    "categorical_features": categorical_features,
    "removed_columns": columns_to_remove,
    "constant_columns": constant_columns,
    "model": best_model_name,
    "training_samples": len(X_train),
    "testing_samples": len(X_test),
    "accuracy": final_accuracy,
    "precision": final_precision,
    "recall": final_recall,
    "f1_score": final_f1,
    "roc_auc": final_roc_auc
}

joblib.dump(
    feature_information,
    os.path.join(
        MODEL_DIR,
        "feature_information.pkl"
    )
)

print(
    "\nFeature information saved."
)


# ================================================================
# 44. LOAD SAVED MODEL AND TEST IT
# ================================================================
#
# This verifies that the saved model actually works.
# ================================================================

print("\n" + "=" * 70)
print("TESTING SAVED MODEL")
print("=" * 70)

loaded_pipeline = joblib.load(
    MODEL_PATH
)

loaded_predictions = (
    loaded_pipeline.predict(X_test)
)

loaded_probabilities = (
    loaded_pipeline.predict_proba(X_test)[:, 1]
)

loaded_accuracy = accuracy_score(
    y_test,
    loaded_predictions
)

print(
    "\nAccuracy from loaded model:",
    round(loaded_accuracy, 4)
)

print(
    "Original accuracy:",
    round(final_accuracy, 4)
)

if abs(
    loaded_accuracy - final_accuracy
) < 0.000001:

    print(
        "\nSUCCESS: Saved model works correctly!"
    )

else:

    print(
        "\nWARNING: Loaded model results differ."
    )


# ================================================================
# 45. TEST A SINGLE STUDENT
# ================================================================
#
# IMPORTANT:
# The feature names below MUST be changed to match your dataset.
#
# This example is only a template.
# ================================================================

print("\n" + "=" * 70)
print("SINGLE STUDENT PREDICTION EXAMPLE")
print("=" * 70)

print(
    "\nYour actual feature columns are:"
)

print(
    X.columns.tolist()
)

print(
    "\nCreate a dictionary using the exact same feature names "
    "when testing a new student."
)

# Example:
#
# new_student = pd.DataFrame([{
#     "cgpa": 8.5,
#     "attendance": 90,
#     "backlogs": 0,
#     "internships": 2,
#     "projects": 3,
#     "certifications": 4,
#     "coding_score": 85,
#     "aptitude_score": 82,
#     "communication_score": 78
# }])
#
# prediction = loaded_pipeline.predict(
#     new_student
# )[0]
#
# probability = loaded_pipeline.predict_proba(
#     new_student
# )[0][1]
#
# if prediction == 1:
#     result = "PLACED"
# else:
#     result = "NOT PLACED"
#
# print("Prediction :", result)
# print("Probability:", round(probability * 100, 2), "%")


# ================================================================
# 46. FINAL PROJECT SUMMARY
# ================================================================

print("\n")
print("=" * 70)
print("PHASE 1 COMPLETED")
print("=" * 70)

print("\nStudent Placement Prediction System")
print("------------------------------------")

print(
    f"Dataset Size        : {len(df)} rows"
)

print(
    f"Input Features      : {X.shape[1]}"
)

print(
    f"Training Samples    : {len(X_train)}"
)

print(
    f"Testing Samples     : {len(X_test)}"
)

print(
    f"Selected Model      : {best_model_name}"
)

print(
    f"Accuracy            : {final_accuracy:.4f}"
)

print(
    f"Precision           : {final_precision:.4f}"
)

print(
    f"Recall              : {final_recall:.4f}"
)

print(
    f"F1 Score            : {final_f1:.4f}"
)

print(
    f"ROC-AUC             : {final_roc_auc:.4f}"
)

print(
    f"\nModel saved at:"
)

print(
    MODEL_PATH
)

print("\nGenerated files:")
print("----------------------------")
print("models/")
print("  ├── placement_pipeline.pkl")
print("  └── feature_information.pkl")
print("")
print("reports/")
print("  ├── model_comparison.csv")
print("  ├── final_metrics.csv")
print("  └── figures/")
print("       ├── target_distribution.png")
print("       ├── numerical_distributions.png")
print("       ├── correlation_matrix.png")
print("       ├── confusion matrices")
print("       ├── roc_curves.png")
print("       ├── final_confusion_matrix.png")
print("       ├── final_roc_curve.png")
print("       └── shap_summary.png")
print("=" * 70)
```
